# SignLearn-08B — Macro-F1-Selected Motion Fusion

This experiment adds the 303 motion/geometry features from SignLearn-07 **without throwing
away the current mirror-robust model**. The base projection and every downstream learned
weight are copied from the deployed 189-feature checkpoint. The new motion projection is
initialized to zero, so the 492-feature model must make the same predictions as the source
model before fine-tuning.

**Protocol**

1. Use Google train signers for training and Google validation signers for selection.
2. Keep Google test rows quarantined; do not inspect test performance in this notebook.
3. Keep WLASL completely external and unused.
4. Warm up only the new projection, then fine-tune the full network at a low learning rate.
5. Evaluate macro F1 after every fine-tuning epoch and save the best macro-F1 checkpoint.
6. Promote only if validation macro F1 improves by at least 0.20 percentage points, accuracy
   is non-inferior, and mirrored-view robustness is preserved.

Enable a Kaggle GPU before running all cells. Attach the existing temporal cache and the
current 189-feature mirror-robust deployment bundle.

```python
from pathlib import Path, PurePosixPath
from IPython.display import display
import hashlib
import json
import shutil
import time
import warnings
import zipfile

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import tensorflow as tf

from sklearn.metrics import accuracy_score, f1_score, top_k_accuracy_score

warnings.filterwarnings('ignore', category=FutureWarning)
sns.set_theme(style='whitegrid', context='notebook')
SEED = 42
BATCH_SIZE = 128
WARMUP_EPOCHS = 5
FINE_TUNE_EPOCHS = 24
AUGMENT_PROBABILITY = 0.50
WARMUP_LEARNING_RATE = 3e-4
FINE_TUNE_LEARNING_RATE = 5e-5
MINIMUM_MACRO_F1_GAIN = 0.002
ACCURACY_TOLERANCE = 0.001
MIRROR_F1_TOLERANCE = 0.005
TARGET_ACCEPTED_ACCURACY = 0.85
MINIMUM_COVERAGE = 0.10

np.random.seed(SEED)
tf.keras.utils.set_random_seed(SEED)
try:
    tf.config.experimental.enable_op_determinism()
except Exception:
    pass

KAGGLE_INPUT = Path('/kaggle/input')
OUTPUT_DIR = Path('/kaggle/working/signlearn_08b')
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
print('TensorFlow:', tf.__version__)
print('GPUs:', tf.config.list_physical_devices('GPU'))
print('Output:', OUTPUT_DIR)
```

## 1. Locate the exact production checkpoint

The resolver reads deployment manifests and accepts only a 189-feature model. It gives
preference to bundles containing the mirror camera policy, preventing the 492-feature
SignLearn-07 candidate from being selected accidentally.

```python
EXTRACTED_SOURCE = OUTPUT_DIR / 'source_bundle'

def manifest_feature_count(manifest):
    shape = manifest.get('input_shape', [])
    return int(shape[-1]) if shape else None

source_candidates = []

# Already-extracted Kaggle datasets.
for manifest_path in KAGGLE_INPUT.rglob('deployment_manifest.json'):
    try:
        manifest = json.loads(manifest_path.read_text(encoding='utf-8'))
    except Exception:
        continue
    model_path = manifest_path.parent / 'signlearn_model.keras'
    if manifest_feature_count(manifest) != 189 or not model_path.exists():
        continue
    evidence = (str(manifest_path) + json.dumps(manifest)).lower()
    score = 20 * ('mirror' in evidence) + 50 * (manifest_path.parent / 'camera_orientation_policy.json').exists()
    source_candidates.append((score, 'directory', manifest_path.parent, manifest))

# Zip bundles uploaded as Kaggle datasets.
for zip_path in KAGGLE_INPUT.rglob('*.zip'):
    try:
        with zipfile.ZipFile(zip_path) as archive:
            names = set(archive.namelist())
            for member in names:
                if not member.endswith('deployment_manifest.json'):
                    continue
                try:
                    manifest = json.loads(archive.read(member).decode('utf-8'))
                except Exception:
                    continue
                parent = PurePosixPath(member).parent
                model_member = str(parent / 'signlearn_model.keras')
                camera_member = str(parent / 'camera_orientation_policy.json')
                if manifest_feature_count(manifest) != 189 or model_member not in names:
                    continue
                evidence = (str(zip_path) + member + json.dumps(manifest)).lower()
                score = 20 * ('mirror' in evidence) + 50 * (camera_member in names)
                source_candidates.append((score, 'zip', (zip_path, str(parent)), manifest))
    except zipfile.BadZipFile:
        continue

if not source_candidates:
    raise FileNotFoundError(
        'No 189-feature deployment model found. Attach signlearn_06_motion_input_v1.zip '
        'or the SignLearn mirror-robust deployment bundle.'
    )

source_candidates.sort(key=lambda item: item[0], reverse=True)
source_score, source_kind, source_location, source_manifest = source_candidates[0]
if source_kind == 'directory':
    SOURCE_ROOT = source_location
else:
    zip_path, member_parent = source_location
    if EXTRACTED_SOURCE.exists():
        shutil.rmtree(EXTRACTED_SOURCE)
    with zipfile.ZipFile(zip_path) as archive:
        archive.extractall(EXTRACTED_SOURCE)
    SOURCE_ROOT = EXTRACTED_SOURCE / Path(member_parent)

SOURCE_MODEL_PATH = SOURCE_ROOT / 'signlearn_model.keras'
LABEL_MAP_PATH = SOURCE_ROOT / 'label_map.json'
FEATURE_NAMES_PATH = SOURCE_ROOT / 'frame_feature_names.json'
TEMPORAL_CONFIG_PATH = SOURCE_ROOT / 'temporal_configuration.json'
SOURCE_MANIFEST_PATH = SOURCE_ROOT / 'deployment_manifest.json'

cache_candidates = [
    path for path in KAGGLE_INPUT.rglob('mvp50_temporal_48_frames.npz')
    if 'wlasl' not in str(path).lower()
]
if not cache_candidates:
    raise FileNotFoundError('Attach the SignLearn-03 mvp50_temporal_48_frames.npz cache.')
CACHE_PATH = cache_candidates[0]

required = [SOURCE_MODEL_PATH, LABEL_MAP_PATH, FEATURE_NAMES_PATH, TEMPORAL_CONFIG_PATH]
missing = [str(path) for path in required if not path.exists()]
if missing:
    raise FileNotFoundError(f'Missing source artifacts: {missing}')
assert 'wlasl' not in str(CACHE_PATH).lower()
print('Source score:', source_score)
print('Source model:', SOURCE_MODEL_PATH)
print('Temporal cache:', CACHE_PATH)
```

```python
cache = np.load(CACHE_PATH, allow_pickle=False)
X = cache['X'].astype(np.float32)
y = cache['y'].astype(np.int32)
splits = cache['splits'].astype(str)
participant_ids = cache['participant_ids']
sequence_ids = cache['sequence_ids']

with LABEL_MAP_PATH.open('r', encoding='utf-8') as stream:
    label_map = json.load(stream)
with FEATURE_NAMES_PATH.open('r', encoding='utf-8') as stream:
    base_feature_names = json.load(stream)

train_mask = splits == 'train'
validation_mask = splits == 'validation'
test_mask = splits == 'test'
X_train, y_train = X[train_mask], y[train_mask]
X_validation, y_validation = X[validation_mask], y[validation_mask]

train_signers = set(participant_ids[train_mask].tolist())
validation_signers = set(participant_ids[validation_mask].tolist())
assert train_signers.isdisjoint(validation_signers)
assert X.shape[1:] == (48, 189)
assert len(base_feature_names) == 189
assert sorted(label_map.values()) == list(range(50))

display(pd.DataFrame({
    'sequences': [len(X_train), len(X_validation), int(test_mask.sum())],
    'signers': [len(train_signers), len(validation_signers),
                len(set(participant_ids[test_mask].tolist()))],
    'status': ['training', 'model selection', 'quarantined — not evaluated'],
}, index=['train', 'validation', 'test']))
print('WLASL loaded: False')
print('Google test predictions computed: False')
```

## 2. Recreate the SignLearn-07 motion/geometry feature contract

```python
BASE_FEATURE_COUNT = 189
MOTION_COORDINATE_COUNT = 144
GEOMETRY_FEATURE_NAMES = [
    'wrist_to_wrist_distance',
    'left_wrist_to_mouth_distance', 'right_wrist_to_mouth_distance',
    'left_wrist_to_nose_distance', 'right_wrist_to_nose_distance',
    'left_wrist_to_shoulder_distance', 'right_wrist_to_shoulder_distance',
    'left_hand_spread', 'right_hand_spread',
    'left_thumb_index_distance', 'right_thumb_index_distance',
    'left_hand_to_pose_wrist_distance', 'right_hand_to_pose_wrist_distance',
    'left_fingertip_to_mouth_min_distance', 'right_fingertip_to_mouth_min_distance',
]
ENHANCED_FEATURE_COUNT = BASE_FEATURE_COUNT + 2 * MOTION_COORDINATE_COUNT + len(GEOMETRY_FEATURE_NAMES)
EXTRA_FEATURE_COUNT = ENHANCED_FEATURE_COUNT - BASE_FEATURE_COUNT
assert ENHANCED_FEATURE_COUNT == 492

reflection_sign = np.ones(BASE_FEATURE_COUNT, dtype=np.float32)
reflection_sign[0:63:3] = -1.0
reflection_sign[63:126:3] = -1.0
reflection_sign[126:144:2] = -1.0
reflection_sign[144:184:2] = -1.0
mirror_permutation = np.arange(BASE_FEATURE_COUNT, dtype=np.int32)
mirror_permutation[0:63] = np.arange(63, 126)
mirror_permutation[63:126] = np.arange(0, 63)
mirror_permutation[184], mirror_permutation[185] = 185, 184
TF_REFLECTION_SIGN = tf.constant(reflection_sign, tf.float32)
TF_MIRROR_PERMUTATION = tf.constant(mirror_permutation, tf.int32)

def mirror_base_features(features):
    return tf.gather(features, TF_MIRROR_PERMUTATION, axis=-1) * TF_REFLECTION_SIGN

def vector_distance(first, second):
    return tf.sqrt(tf.reduce_sum(tf.square(first - second), axis=-1) + 1e-8)

def build_enhanced_features(features):
    features = tf.cast(features, tf.float32)
    motion_coordinates = features[:, :MOTION_COORDINATE_COUNT]
    detection = features[:, 184:189]

    previous_coordinates = tf.concat([motion_coordinates[:1], motion_coordinates[:-1]], axis=0)
    velocity = motion_coordinates - previous_coordinates
    previous_detection = tf.concat([detection[:1], detection[:-1]], axis=0)
    velocity_masks = tf.concat([
        tf.repeat((detection[:, 0:1] > 0.5) & (previous_detection[:, 0:1] > 0.5), 63, axis=1),
        tf.repeat((detection[:, 1:2] > 0.5) & (previous_detection[:, 1:2] > 0.5), 63, axis=1),
        tf.repeat((detection[:, 2:3] > 0.5) & (previous_detection[:, 2:3] > 0.5), 18, axis=1),
    ], axis=1)
    velocity = tf.where(velocity_masks, velocity, 0.0)
    previous_velocity = tf.concat([velocity[:1], velocity[:-1]], axis=0)
    acceleration = velocity - previous_velocity
    previous_velocity_mask = tf.concat([velocity_masks[:1], velocity_masks[:-1]], axis=0)
    acceleration = tf.where(velocity_masks & previous_velocity_mask, acceleration, 0.0)
    velocity = tf.clip_by_value(velocity, -5.0, 5.0)
    acceleration = tf.clip_by_value(acceleration, -5.0, 5.0)

    left_hand = tf.reshape(features[:, 0:63], [-1, 21, 3])
    right_hand = tf.reshape(features[:, 63:126], [-1, 21, 3])
    pose = tf.reshape(features[:, 126:144], [-1, 9, 2])
    lips = tf.reshape(features[:, 144:184], [-1, 20, 2])
    left_wrist, right_wrist = left_hand[:, 0, :2], right_hand[:, 0, :2]
    nose, left_shoulder, right_shoulder = pose[:, 0], pose[:, 1], pose[:, 2]
    left_pose_wrist, right_pose_wrist = pose[:, 5], pose[:, 6]
    mouth = tf.reduce_mean(lips, axis=1)
    fingertip_indices = tf.constant([4, 8, 12, 16, 20], tf.int32)
    left_tips = tf.gather(left_hand[:, :, :2], fingertip_indices, axis=1)
    right_tips = tf.gather(right_hand[:, :, :2], fingertip_indices, axis=1)
    left_spread = tf.reduce_mean(vector_distance(left_tips, left_wrist[:, None, :]), axis=1)
    right_spread = tf.reduce_mean(vector_distance(right_tips, right_wrist[:, None, :]), axis=1)
    left_tip_to_mouth = tf.reduce_min(vector_distance(left_tips, mouth[:, None, :]), axis=1)
    right_tip_to_mouth = tf.reduce_min(vector_distance(right_tips, mouth[:, None, :]), axis=1)

    geometry = tf.stack([
        vector_distance(left_wrist, right_wrist),
        vector_distance(left_wrist, mouth), vector_distance(right_wrist, mouth),
        vector_distance(left_wrist, nose), vector_distance(right_wrist, nose),
        vector_distance(left_wrist, left_shoulder), vector_distance(right_wrist, right_shoulder),
        left_spread, right_spread,
        vector_distance(left_hand[:, 4, :2], left_hand[:, 8, :2]),
        vector_distance(right_hand[:, 4, :2], right_hand[:, 8, :2]),
        vector_distance(left_wrist, left_pose_wrist), vector_distance(right_wrist, right_pose_wrist),
        left_tip_to_mouth, right_tip_to_mouth,
    ], axis=1)
    left_valid = detection[:, 0] > 0.5
    right_valid = detection[:, 1] > 0.5
    pose_valid = detection[:, 2] > 0.5
    lip_valid = detection[:, 3] > 0.5
    geometry_masks = tf.stack([
        left_valid & right_valid,
        left_valid & lip_valid, right_valid & lip_valid,
        left_valid & pose_valid, right_valid & pose_valid,
        left_valid & pose_valid, right_valid & pose_valid,
        left_valid, right_valid, left_valid, right_valid,
        left_valid & pose_valid, right_valid & pose_valid,
        left_valid & lip_valid, right_valid & lip_valid,
    ], axis=1)
    geometry = tf.where(geometry_masks, tf.clip_by_value(geometry, 0.0, 10.0), 0.0)
    enhanced = tf.concat([features, velocity, acceleration, geometry], axis=1)
    return tf.ensure_shape(enhanced, [48, ENHANCED_FEATURE_COUNT])

enhanced_feature_names = (
    base_feature_names
    + [f'velocity_{name}' for name in base_feature_names[:MOTION_COORDINATE_COUNT]]
    + [f'acceleration_{name}' for name in base_feature_names[:MOTION_COORDINATE_COUNT]]
    + GEOMETRY_FEATURE_NAMES
)
sample = tf.convert_to_tensor(X_validation[0])
enhanced_sample = build_enhanced_features(sample)
np.testing.assert_allclose(
    mirror_base_features(mirror_base_features(sample)).numpy(), sample.numpy(), atol=1e-6
)
assert enhanced_sample.shape == (48, 492)
assert bool(tf.reduce_all(tf.math.is_finite(enhanced_sample)))
assert np.allclose(enhanced_sample.numpy()[0, 189:477], 0.0)
assert len(enhanced_feature_names) == 492
print('Feature and mirror contract: OK')
```

## 3. Create deterministic training and validation pipelines

```python
AUTOTUNE = tf.data.AUTOTUNE

def randomly_mirror(features, label):
    features = tf.cast(features, tf.float32)
    mirrored = mirror_base_features(features)
    use_mirror = tf.random.uniform([], seed=SEED) < AUGMENT_PROBABILITY
    return tf.cond(use_mirror, lambda: mirrored, lambda: features), label

def add_enhanced_features(features, label):
    return build_enhanced_features(features), label

enhanced_train = (
    tf.data.Dataset.from_tensor_slices((X_train, y_train))
    .shuffle(len(X_train), seed=SEED, reshuffle_each_iteration=True)
    .map(randomly_mirror, num_parallel_calls=AUTOTUNE, deterministic=True)
    .map(add_enhanced_features, num_parallel_calls=AUTOTUNE, deterministic=True)
    .batch(BATCH_SIZE).prefetch(AUTOTUNE)
)

def evaluation_dataset(features, labels, enhanced=False, mirrored=False):
    dataset = tf.data.Dataset.from_tensor_slices((features, labels))
    if mirrored:
        dataset = dataset.map(
            lambda item, label: (mirror_base_features(tf.cast(item, tf.float32)), label),
            num_parallel_calls=AUTOTUNE, deterministic=True,
        )
    if enhanced:
        dataset = dataset.map(add_enhanced_features, num_parallel_calls=AUTOTUNE, deterministic=True)
    return dataset.batch(BATCH_SIZE).prefetch(AUTOTUNE)

base_validation = evaluation_dataset(X_validation, y_validation)
base_validation_mirrored = evaluation_dataset(X_validation, y_validation, mirrored=True)
enhanced_validation = evaluation_dataset(X_validation, y_validation, enhanced=True)
enhanced_validation_mirrored = evaluation_dataset(
    X_validation, y_validation, enhanced=True, mirrored=True
)
print('Training input:', next(iter(enhanced_train))[0].shape)
```

## 4. Load and audit the current mirror-robust model

```python
def predict_dataset(model, dataset):
    features_only = dataset.map(lambda features, labels: features)
    return model.predict(features_only, verbose=0)

def probability_metrics(true_labels, probabilities):
    predictions = probabilities.argmax(axis=1)
    return {
        'accuracy': float(accuracy_score(true_labels, predictions)),
        'macro_f1': float(f1_score(true_labels, predictions, average='macro')),
        'top5_accuracy': float(top_k_accuracy_score(
            true_labels, probabilities, k=5, labels=np.arange(50)
        )),
    }

source_model = tf.keras.models.load_model(SOURCE_MODEL_PATH, compile=False)
assert tuple(source_model.input_shape[1:]) == (48, 189)
source_validation_probabilities = predict_dataset(source_model, base_validation)
source_mirrored_probabilities = predict_dataset(source_model, base_validation_mirrored)
source_metrics = probability_metrics(y_validation, source_validation_probabilities)
source_mirrored_metrics = probability_metrics(y_validation, source_mirrored_probabilities)
display(pd.DataFrame([
    {'model': 'source_189', 'view': 'original', **source_metrics},
    {'model': 'source_189', 'view': 'mirrored', **source_mirrored_metrics},
]).style.format({'accuracy': '{:.4f}', 'macro_f1': '{:.4f}', 'top5_accuracy': '{:.4f}'}))
```

## 5. Build the transfer-initialized fusion model

The base branch computes the original 189-to-128 projection. The new 303-to-128 motion
projection has no bias and starts at exactly zero. Both projections are added before the
original ReLU, making the model functionally identical to the source at initialization.

```python
def build_transfer_model():
    inputs = tf.keras.Input((48, 492), dtype=tf.float32, name='enhanced_landmark_sequence')
    noisy = tf.keras.layers.GaussianNoise(0.015, name='landmark_noise')(inputs)
    base_inputs = noisy[:, :, :BASE_FEATURE_COUNT]
    motion_inputs = noisy[:, :, BASE_FEATURE_COUNT:]
    base_projection = tf.keras.layers.Dense(
        128, activation=None, name='base_projection'
    )(base_inputs)
    motion_projection = tf.keras.layers.Dense(
        128, activation=None, use_bias=False, kernel_initializer='zeros',
        kernel_regularizer=tf.keras.regularizers.l2(1e-4), name='motion_projection'
    )(motion_inputs)
    x = tf.keras.layers.Add(name='projection_fusion')([base_projection, motion_projection])
    x = tf.keras.layers.Activation('relu', name='frame_projection_activation')(x)
    x = tf.keras.layers.LayerNormalization(name='frame_normalization')(x)
    x = tf.keras.layers.SpatialDropout1D(0.15, name='spatial_dropout')(x)
    x = tf.keras.layers.Bidirectional(
        tf.keras.layers.LSTM(96, return_sequences=True, dropout=0.20), name='bilstm_1'
    )(x)
    x = tf.keras.layers.Bidirectional(
        tf.keras.layers.LSTM(64, return_sequences=True, dropout=0.20), name='bilstm_2'
    )(x)
    attention = tf.keras.layers.MultiHeadAttention(
        num_heads=4, key_dim=32, dropout=0.15, name='temporal_self_attention'
    )(x, x)
    x = tf.keras.layers.LayerNormalization(name='attention_residual')(
        tf.keras.layers.Add(name='attention_add')([x, attention])
    )
    average_pool = tf.keras.layers.GlobalAveragePooling1D(name='average_pool')(x)
    maximum_pool = tf.keras.layers.GlobalMaxPooling1D(name='maximum_pool')(x)
    x = tf.keras.layers.Concatenate(name='pooled_features')([average_pool, maximum_pool])
    x = tf.keras.layers.Dense(
        192, activation='relu', kernel_initializer='he_normal',
        kernel_regularizer=tf.keras.regularizers.l2(2e-5), name='classification_hidden'
    )(x)
    x = tf.keras.layers.Dropout(0.35, name='classification_dropout')(x)
    outputs = tf.keras.layers.Dense(50, activation='softmax', name='sign_probabilities')(x)
    return tf.keras.Model(inputs, outputs, name='signlearn_transfer_motion_fusion')

transfer_model = build_transfer_model()

source_projection = source_model.get_layer('frame_projection')
transfer_model.get_layer('base_projection').set_weights(source_projection.get_weights())
assert np.allclose(transfer_model.get_layer('motion_projection').get_weights()[0], 0.0)

for layer_name in [
    'frame_normalization', 'bilstm_1', 'bilstm_2',
    'temporal_self_attention', 'attention_residual', 'sign_probabilities',
]:
    transfer_model.get_layer(layer_name).set_weights(source_model.get_layer(layer_name).get_weights())

source_hidden = [
    layer for layer in source_model.layers
    if isinstance(layer, tf.keras.layers.Dense) and layer.units == 192
]
assert len(source_hidden) == 1
transfer_model.get_layer('classification_hidden').set_weights(source_hidden[0].get_weights())
print('Transferred source weights into the 492-feature fusion model.')
```

```python
probe_base = tf.convert_to_tensor(X_validation[:64], tf.float32)
probe_enhanced = tf.map_fn(
    build_enhanced_features, probe_base,
    fn_output_signature=tf.TensorSpec((48, 492), tf.float32),
)
source_probe = source_model(probe_base, training=False).numpy()
transfer_probe = transfer_model(probe_enhanced, training=False).numpy()
INITIALIZATION_MAX_ABS_ERROR = float(np.max(np.abs(source_probe - transfer_probe)))
print('Maximum prediction difference at initialization:', INITIALIZATION_MAX_ABS_ERROR)
assert INITIALIZATION_MAX_ABS_ERROR < 1e-5, (
    'Transfer initialization is not equivalent to the source model. Stop before training.'
)
print('Exact-behavior initialization check: PASSED')
```

## 6. Stage A — warm up only the new motion projection

```python
def compile_model(model, learning_rate):
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate, clipnorm=1.0),
        loss=tf.keras.losses.SparseCategoricalCrossentropy(),
        metrics=[
            tf.keras.metrics.SparseCategoricalAccuracy(name='accuracy'),
            tf.keras.metrics.SparseTopKCategoricalAccuracy(k=5, name='top5_accuracy'),
        ],
    )

for layer in transfer_model.layers:
    layer.trainable = layer.name == 'motion_projection'
compile_model(transfer_model, WARMUP_LEARNING_RATE)
print('Trainable layers:', [layer.name for layer in transfer_model.layers if layer.trainable])

WARMUP_PATH = OUTPUT_DIR / 'best_motion_warmup.keras'
warmup_started = time.time()
warmup_history = transfer_model.fit(
    enhanced_train,
    validation_data=enhanced_validation,
    epochs=WARMUP_EPOCHS,
    callbacks=[
        tf.keras.callbacks.ModelCheckpoint(
            WARMUP_PATH, monitor='val_loss', save_best_only=True, verbose=1
        )
    ],
    verbose=2,
)
warmup_frame = pd.DataFrame(warmup_history.history)
warmup_frame['stage'] = 'motion_projection_warmup'
warmup_frame.to_csv(OUTPUT_DIR / 'warmup_history.csv', index=False)
print(f'Warm-up minutes: {(time.time() - warmup_started) / 60:.1f}')
```

## 7. Stage B — macro-F1-selected full-network fine-tuning

SignLearn-08 saved the lowest-loss epoch even though promotion was judged by macro F1.
This corrected callback computes validation macro F1 after every epoch, saves that exact
checkpoint, and applies early stopping to the same selection metric.

```python
transfer_model = tf.keras.models.load_model(WARMUP_PATH, compile=False)
for layer in transfer_model.layers:
    layer.trainable = True
compile_model(transfer_model, FINE_TUNE_LEARNING_RATE)

BEST_MODEL_PATH = OUTPUT_DIR / 'best_macro_f1_transfer_motion_492.keras'

class ValidationMacroF1Checkpoint(tf.keras.callbacks.Callback):
    def __init__(self, validation_dataset, true_labels, filepath, patience=7, min_delta=1e-5):
        super().__init__()
        self.validation_dataset = validation_dataset
        self.true_labels = np.asarray(true_labels)
        self.filepath = filepath
        self.patience = patience
        self.min_delta = min_delta
        self.best = -np.inf
        self.best_epoch = None
        self.wait = 0
        self.records = []

    def on_epoch_end(self, epoch, logs=None):
        logs = logs if logs is not None else {}
        probabilities = predict_dataset(self.model, self.validation_dataset)
        predictions = probabilities.argmax(axis=1)
        macro_f1 = float(f1_score(self.true_labels, predictions, average='macro'))
        accuracy = float(accuracy_score(self.true_labels, predictions))
        logs['val_macro_f1_external'] = macro_f1
        logs['val_accuracy_external'] = accuracy
        self.records.append({
            'epoch': int(epoch + 1),
            'val_macro_f1_external': macro_f1,
            'val_accuracy_external': accuracy,
        })
        print(
            f' — external val_macro_f1: {macro_f1:.6f}'
            f' — external val_accuracy: {accuracy:.6f}'
        )
        if macro_f1 > self.best + self.min_delta:
            self.best = macro_f1
            self.best_epoch = int(epoch + 1)
            self.wait = 0
            self.model.save(self.filepath, overwrite=True)
            print(f' — saved new best macro-F1 checkpoint: {self.filepath}')
        else:
            self.wait += 1
            if self.wait >= self.patience:
                self.model.stop_training = True
                print(
                    f' — early stopping: validation macro F1 did not improve for '
                    f'{self.patience} epochs'
                )

macro_f1_checkpoint = ValidationMacroF1Checkpoint(
    enhanced_validation, y_validation, BEST_MODEL_PATH, patience=7
)
callbacks = [
    macro_f1_checkpoint,
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss', factor=0.5, patience=3, min_lr=1e-6, verbose=1
    ),
]
fine_tune_started = time.time()
fine_tune_history = transfer_model.fit(
    enhanced_train,
    validation_data=enhanced_validation,
    epochs=FINE_TUNE_EPOCHS,
    callbacks=callbacks,
    verbose=2,
)
fine_tune_frame = pd.DataFrame(fine_tune_history.history)
fine_tune_frame['stage'] = 'full_fine_tuning'
fine_tune_frame.to_csv(OUTPUT_DIR / 'fine_tune_history.csv', index=False)
epoch_macro_frame = pd.DataFrame(macro_f1_checkpoint.records)
epoch_macro_frame.to_csv(OUTPUT_DIR / 'epoch_validation_macro_f1.csv', index=False)
print('Best validation macro F1 epoch:', macro_f1_checkpoint.best_epoch)
print('Best validation macro F1:', macro_f1_checkpoint.best)
print(f'Fine-tuning minutes: {(time.time() - fine_tune_started) / 60:.1f}')
```

```python
candidate_model = tf.keras.models.load_model(BEST_MODEL_PATH, compile=False)
candidate_validation_probabilities = predict_dataset(candidate_model, enhanced_validation)
candidate_mirrored_probabilities = predict_dataset(candidate_model, enhanced_validation_mirrored)
candidate_metrics = probability_metrics(y_validation, candidate_validation_probabilities)
candidate_mirrored_metrics = probability_metrics(y_validation, candidate_mirrored_probabilities)

validation_comparison = pd.DataFrame([
    {'model': 'source_189', 'view': 'original', **source_metrics},
    {'model': 'source_189', 'view': 'mirrored', **source_mirrored_metrics},
    {'model': 'transfer_motion_492', 'view': 'original', **candidate_metrics},
    {'model': 'transfer_motion_492', 'view': 'mirrored', **candidate_mirrored_metrics},
])
validation_comparison.to_csv(OUTPUT_DIR / 'validation_transfer_comparison.csv', index=False)
display(validation_comparison.style.format({
    'accuracy': '{:.4f}', 'macro_f1': '{:.4f}', 'top5_accuracy': '{:.4f}'
}))

macro_f1_gain = candidate_metrics['macro_f1'] - source_metrics['macro_f1']
accuracy_gain = candidate_metrics['accuracy'] - source_metrics['accuracy']
mirror_f1_gain = candidate_mirrored_metrics['macro_f1'] - source_mirrored_metrics['macro_f1']
PROMOTE_CANDIDATE = bool(
    macro_f1_gain >= MINIMUM_MACRO_F1_GAIN
    and accuracy_gain >= -ACCURACY_TOLERANCE
    and mirror_f1_gain >= -MIRROR_F1_TOLERANCE
)

source_class_f1 = f1_score(
    y_validation, source_validation_probabilities.argmax(axis=1),
    labels=np.arange(50), average=None, zero_division=0,
)
candidate_class_f1 = f1_score(
    y_validation, candidate_validation_probabilities.argmax(axis=1),
    labels=np.arange(50), average=None, zero_division=0,
)
index_to_sign = {int(index): sign for sign, index in label_map.items()}
per_class = pd.DataFrame({
    'class_index': np.arange(50),
    'sign': [index_to_sign[index] for index in range(50)],
    'source_f1': source_class_f1,
    'candidate_f1': candidate_class_f1,
    'f1_delta': candidate_class_f1 - source_class_f1,
}).sort_values('f1_delta', ascending=False)
per_class.to_csv(OUTPUT_DIR / 'validation_per_class_transfer_delta.csv', index=False)
print('Original-view macro F1 gain:', f'{macro_f1_gain:+.4f}')
print('Original-view accuracy gain:', f'{accuracy_gain:+.4f}')
print('Mirrored-view macro F1 gain:', f'{mirror_f1_gain:+.4f}')
print('PROMOTE CANDIDATE:', PROMOTE_CANDIDATE)
display(pd.concat([per_class.head(8), per_class.tail(8)]))
```

```python
def selective_curve(probabilities, true_labels, thresholds):
    predictions = probabilities.argmax(axis=1)
    confidences = probabilities.max(axis=1)
    rows = []
    for threshold in thresholds:
        accepted = confidences >= threshold
        rows.append({
            'threshold': float(threshold),
            'coverage': float(accepted.mean()),
            'accepted_accuracy': (
                float((predictions[accepted] == true_labels[accepted]).mean())
                if accepted.any() else np.nan
            ),
            'accepted_count': int(accepted.sum()),
        })
    return pd.DataFrame(rows)

candidate_selective = selective_curve(
    candidate_validation_probabilities, y_validation, np.arange(0.0, 1.0, 0.01)
)
eligible = candidate_selective[
    (candidate_selective['accepted_accuracy'] >= TARGET_ACCEPTED_ACCURACY)
    & (candidate_selective['coverage'] >= MINIMUM_COVERAGE)
]
if eligible.empty:
    selected_threshold_row = candidate_selective.sort_values(
        ['accepted_accuracy', 'coverage'], ascending=False
    ).iloc[0]
else:
    selected_threshold_row = eligible.sort_values('threshold').iloc[0]
CANDIDATE_CONFIDENCE_THRESHOLD = float(selected_threshold_row['threshold'])
candidate_selective.to_csv(OUTPUT_DIR / 'candidate_validation_selective_curve.csv', index=False)
display(selected_threshold_row.to_frame('candidate_validation_policy'))
```
